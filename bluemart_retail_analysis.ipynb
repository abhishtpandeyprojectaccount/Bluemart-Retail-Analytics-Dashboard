### Data analysis and visualization

### 1. How are sales changing over time?

```python
monthly_sales = df.groupby(df['date'].dt.to_period('M'))['total_value'].sum()
monthly_sales.index = monthly_sales.index.astype(str)

# Visualization -- Line Chart
plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index, monthly_sales.values,marker='o',color='purple')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

#### 2 Which categories generate the highest sales and profit ?

```python
# profit = Total value-Quantity*Cost price   ---- formula ----
```

```python
df['profit'] = df['total_value'] - (df['quantity'] * df['cost_price'])

category_perf=df.groupby('category').agg(
    total_sales=('total_value', 'sum'),
    total_profit=('profit', 'sum')).sort_values('total_sales' ,ascending=False)

print(category_perf)
# Visualization -- bar chart

x=np.arange(len(category_perf)) #       [0,1,2,3,4,5,6]
width=0.35

plt.figure(figsize=(11, 5))
plt.bar(x-width/2,category_perf['total_sales'],width=width,label='Total_Sales', color='blue')
plt.bar(x+width/2,category_perf['total_profit'],width=width,label='Total_Profit', color='Green')
plt.xticks(x, category_perf.index, rotation=90)
plt.ylabel('Amount')
plt.xlabel('Category')
plt.title('Category-wise Sales vs Profit')
plt.legend()
plt.tight_layout()
plt.show()
```

**3. Which products are the most profitable?**

```python
product_pref=df.groupby('subcategory').agg(total_sales=('total_value', 'sum'),
                                            total_profit=('profit', 'sum'))
print(product_pref)

product_pref = product_pref.sort_values('total_profit', ascending=False)
top10_profitable=product_pref.head(10)

plt.figure(figsize=(11, 5))
plt.bar(top10_profitable.index, top10_profitable['total_profit'],color='red')
plt.xlabel('Product')
plt.ylabel('Total Profit')
plt.title('Top 10 Profitable Products')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

### AD - HOC Analysis & Final insights

#### Q1.selling products are at risk of stockout (low inventory)?

```python
df.head(5)
```

```python
product_sales = df.groupby('sku_id')['quantity'].sum().reset_index()
product_sales
```

```python
risk_df = product_sales.merge(idf, on='sku_id', how='left')
risk_df
```

```python
high_sales_low_stocks = risk_df[
    (risk_df['quantity'] > risk_df['quantity'].median()) &
    (risk_df['stock_on_hand'] < risk_df['stock_on_hand'].median())
]
```

#### High Sales+Low Stock--> Potential Stockout Risk

```python
high_sales_low_stocks
```

#### Q3 Which products have high sales but low profits?

```python
product_summary = df.groupby('sku_name'). agg(total_sales=('total_value', 'sum'),total_profit=('profit', 'sum'))
product_summary
```

```python
high_sales_low_profit = product_summary.loc[
    (product_summary['total_sales'] > product_summary['total_sales'].median()) &
    (product_summary['total_profit'] < product_summary['total_profit'].median())
]
high_sales_low_profit
```
